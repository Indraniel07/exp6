```python
#5. given year is a leap year or not

a = int(input("Enter the year :"))

if (a % 4 == 0 and a % 100 != 0) or (a % 400 == 0) :
  print(a,"is a leap year")
else :
  print(a,"is not a leap year")
```

```python
#6. increment of given date

d = input("Enter the data in DD-MM-YYYY format :")

day,month,year = map(int,d.split("-"))

if (year % 400 == 0) or (year % 4 == 0 and year % 100 != 0) :
  feb_days = 29
else :
  feb_days = 28

daysinmonth = {
    1: 31,
    2: feb_days,
    3: 31,
    4: 30,
    5: 31,
    6: 30,
    7: 31,
    8: 31,
    9: 30,
    10: 31,
    11: 30,
    12: 31
}

if (month < 1 or month > 12) :
  print("Date  is invalid.")

elif (day < 1 or day > daysinmonth[month]) :
  print("Date  is invalid.")

day = day + 1

if (day > daysinmonth[month]) :
  day = 1
  month = month + 1

if (month > 12) :
  month = 1
  year = year + 1

print("Next date is:",day,"-",month,"-",year)
```

```python
#8. write a python program to calculate the gross salary of an employee

basic = float(input("Enter basic salary: "))

if (basic <= 10000) :
  hra = basic * 0.2
  da = basic * 0.8
elif (basic <= 20000) :
  hra = basic * 0.25
  da = basic * 0.9
else :
  hra = basic * 0.3
  da = basic * 0.95

salary = hra + da + basic

print("Gross salary:",salary)
```

```python
#9. write a python program to calculate income tax for based on annual income :

income = float(input("Enter annual income: "))

if (income <= 250000) :
  tax = 0
elif (income > 250000 and income <= 500000) :
  tax = (income - 250000) * 0.05
elif (income > 500000 and income <= 1000000) :
  tax = (income - 500000) * 0.2 + 250000 * 0.05
elif (income > 1000000) :
  tax = (income - 100000) * 0.3 + 500000 * 0.2 + 250000 * 0.05

print("Income Tax: ",tax)
```
